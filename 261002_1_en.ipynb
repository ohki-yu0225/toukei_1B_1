## Population and Sample
Population: the entire set of data that is the subject of a statistical survey <br>
Sample: the part of the population that is actually surveyed (a subset of the data) <br>
Sampling: extracting a sample from a population <br>

In this exercise, we use sample data as the population and practice extracting samples under several conditions. In practice, the sampled data itself is what we analyze, so we never extract a sample from population data; instead, the population data is unavailable, and we need to consider an appropriate sampling method.

### Random sampling
Random sampling is one way to select a sample from a population without bias. In random sampling, every data point in the population has the same probability of being selected for the sample.

```python
# Randomly select 20 students as a sample
df_r = df.sample(n=20, random_state=0) #randomly extract 20 rows from 'df'
df_r
```

Compare the mean of the randomly selected sample with the mean of the census (the population).


```python
# Calculate the mean of each test for the random sample and the census
mean_r = df_r[['Japanese', 'Math', 'English']].mean() #mean of the random sample
mean_all = df[['Japanese', 'Math', 'English']].mean() #mean of the census

pd.DataFrame([mean_all, mean_r], index=['Census', 'Random sample'])
```

### Sampling and bias

Using the course information, extract only the students in the Science and Mathematics course, or only those in the International course, as a sample.

```python
# Extract students in the Science and Mathematics course as a sample
df_s = df[df['Course'] == 'Science'] #extract rows where the 'Course' column is 'Science'
df_s
```

```python
# Extract students in the International course as a sample
df_i = df[df['Course'] == 'International'] #extract rows where the 'Course' column is 'International'
df_i
```

```python
# Calculate the mean of each test for the Science and International course students, and compare with the random sample and the census
mean_s = df_s[['Japanese', 'Math', 'English']].mean() #mean of the Science and Mathematics course
mean_i = df_i[['Japanese', 'Math', 'English']].mean() #mean of the International course

pd.DataFrame([mean_all, mean_r, mean_s, mean_i], index=['Census', 'Random sample', 'Science', 'International'])
```
