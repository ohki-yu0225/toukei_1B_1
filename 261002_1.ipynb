## 母集団と標本
母集団：統計調査の対象となるデータ全体の集合 <br>
標本：実際に調査対象となる母集団の一部（部分集合データ） <br>
標本抽出：母集団から標本を取り出すこと <br>

今回の実習では，サンプルデータを母集団として，いくつかの条件で標本を抽出する練習をする。実際は，標本抽出されたデータがデータ分析の対象となるので，母集団のデータから標本を抽出するということはなく，母集団のデータがない状態となる適切な標本抽出の方法を検討する必要がある。

### 無作為抽出
母集団から標本を偏りなく選ぶ方法の1つに，無作為抽出がある。無作為抽出では，母集団のどのデータも同じ確率で標本に選ばれる。

```python
# 無作為に20人を標本として抽出する
df_r = df.sample(n=20, random_state=0) #'df'から無作為に20行のデータを抽出
df_r
```

無作為抽出した標本の平均値と，全数調査（母集団）の平均値を比べる。


```python
# 無作為抽出と全数調査の各テストの平均値を計算する
mean_r = df_r[['国語', '数学', '英語']].mean() #無作為抽出の平均値
mean_all = df[['国語', '数学', '英語']].mean() #全数調査の平均値

pd.DataFrame([mean_all, mean_r], index=['全数調査', '無作為抽出'])
```

### 標本抽出とバイアス

クラスの情報を用いて，理数科，または国際科に所属している学生のみを標本として抽出する。

```python
# 理数科の学生を標本として抽出する
df_s = df[df['クラス'] == '理数'] #'クラス'の列が'理数'のデータを抽出
df_s
```

```python
# 国際科の学生を標本として抽出する
df_i = df[df['クラス'] == '国際'] #'クラス'の列が'国際'のデータを抽出
df_i
```

```python
# 理数科・国際科の学生の各テストの平均値を計算し，無作為抽出・全数調査の結果と比較する
mean_s = df_s[['国語', '数学', '英語']].mean() #理数科の平均値
mean_i = df_i[['国語', '数学', '英語']].mean() #国際科の平均値

pd.DataFrame([mean_all, mean_r, mean_s, mean_i], index=['全数調査', '無作為抽出', '理数科', '国際科'])
```
